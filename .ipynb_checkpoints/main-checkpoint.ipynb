## Numerical-Numerical correlation

```python
numeric_cols = ["monthly_rent_npr", "area_sqft_approx", "road_width_ft", "bedrooms",
                "bathrooms", "living_rooms", "kitchens", "total_rooms", "built_up_sqft",
                "no_of_flats", "ward"]

corr_matrix = data[numeric_cols].corr()

plt.figure(figsize=(9, 7))
sns.heatmap(corr_matrix, annot=True, cmap="coolwarm", center=0)
plt.title("Correlation Heatmap — Numeric Features")
plt.tight_layout()
plt.show()

print(corr_matrix["monthly_rent_npr"].sort_values(ascending=False))
```

## bathrooms and bedrooms are the strongest numeric predictors of rent bigger, better-equipped homes cost more. area_sqft_approx and road_width_ft have weaker but still positive correlation. ward is essentially uncorrelated (~-0.11), which makes sense

```python
fig, axes = plt.subplots(1, 2, figsize=(13, 5))

sns.scatterplot(data=data, x="bedrooms", y="monthly_rent_npr", ax=axes[0])
axes[0].set_title("Bedrooms vs Monthly Rent")

sns.scatterplot(data=data, x="area_sqft_approx", y="monthly_rent_npr", ax=axes[1])
axes[1].set_title("Area (sqft) vs Monthly Rent")

plt.tight_layout()
plt.show()
```

## less bedroom house cost less and vice-versa.

## numerical-categorical relationship

```python
print(data.groupby("property_type")["monthly_rent_npr"].agg(["mean","median","count"]).round(0))

plt.figure(figsize=(8, 5))
sns.boxplot(data=data, x="property_type", y="monthly_rent_npr")
plt.title("Monthly Rent by Property Type")
plt.tight_layout()
plt.show()
```

```python



sns.barplot(
    data=data,
    x="property_type",
    y="monthly_rent_npr",
    estimator="mean",
    errorbar=None
)
plt.title(" Monthly Rent by Property Type")
plt.xlabel("Property Type")
plt.ylabel("Monthly Rent (NPR)")

plt.tight_layout()
plt.show()
```

```python
# Keep locations with 3 or more listings
counts = data["location"].value_counts()
data2 = data[data["location"].isin(counts[counts >= 3].index)]

# Median rent by location
median_rent = data2.groupby("location")["monthly_rent_npr"].median().sort_values()

# Bar graph
median_rent.plot(kind="barh", figsize=(10, 6))

plt.title("Median Monthly Rent by Location")
plt.xlabel("Median Monthly Rent (NPR)")
plt.ylabel("Location")
plt.tight_layout()
plt.show()
```

## most expensive area is budhanilkantha and rent around tokha is comparatively less


```python
print(data["city"].value_counts())

plt.figure(figsize=(8, 5))
order = (data.groupby("city")["monthly_rent_npr"]
         .median()
         .sort_values(ascending=False)
         .index)

sns.boxplot(data=data, x="city", y="monthly_rent_npr", order=order)
plt.title("Monthly Rent by City")
plt.xlabel("City")
plt.ylabel("Monthly Rent (NPR)")
plt.tight_layout()
plt.show()

print(data.groupby("city")["monthly_rent_npr"].agg(["median","mean","count"]).sort_values("median", ascending=False))
```

## Kathmandu has the highest and most widely varied rental prices, including several high-rent properties. Lalitpur has relatively higher and more consistent rents, while Bhaktapur generally has the lowest monthly rents. Overall, Kathmandu appears to be the most expensive and variable rental market among the three cities.

## Outlier Detection

```python
Q1 = data["monthly_rent_npr"].quantile(0.25)
Q3 = data["monthly_rent_npr"].quantile(0.75)

IQR = Q3 - Q1

outliers = data[
    (data["monthly_rent_npr"] < Q1 - 1.5 * IQR) | (data["monthly_rent_npr"] > Q3 + 1.5 * IQR)]

print("Number of outliers:", len(outliers))
print(outliers[["title", "monthly_rent_npr", "property_type"]])
```

```python
plt.figure(figsize=(8, 5))

sns.boxplot(x=data["monthly_rent_npr"])
sns.stripplot(x=outliers["monthly_rent_npr"], color="red", size=6)

plt.title("Monthly Rent with Outliers")
plt.xlabel("Monthly Rent (NPR)")
plt.show()
```

## Feature Engineering

```python
data["price_per_sqft"] = data["monthly_rent_npr"] / data["area_sqft_approx"]
data["rent_per_bedroom"] = data["monthly_rent_npr"] / data["bedrooms"]

print(data["price_per_sqft"].describe())
```

```python
data.info()
data.shape
```

```python
data = data.drop(columns=["title", "description", "built_up_sqft", "total_rooms"])
print("Shape after feature engineering:", data.shape)
print(data.columns.tolist())
```

```python
print("Unique locations:", data["location"].nunique(), "out of", len(data), "rows")
print(data["location"].value_counts().head(10))
```

## The location feature contains multiple categories, making it a categorical variable with relatively high cardinality. One-hot encoding can be used to convert these locations into numerical features for the model
