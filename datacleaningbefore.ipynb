```python
import pandas as pd
df =pd.read_csv('nepal_rental_dataset.csv')
df
```

```python
df = df[df["district"] == "Kathmandu"].copy()
print("After filtering to Kathmandu district:", df.shape)
```

```python
df.columns
```

```python
drop_cols = [
    "phone", "email", "url", "images", "agent_or_owner",  # not predictive
    "num_images",           # rarely meaningful for rent prediction
    "listing_date", "status",  #  not property characteristics
    "price_raw", "land_area", "road_size",  #  already parsed into numeric cols
    "province",             
]
```

```python
df = df.drop(columns=[c for c in drop_cols if c in df.columns])
print("After dropping noise columns:", df.shape)
print("\nRemaining columns:", list(df.columns))
```

```python
# Drop rows with no rent value at all (can't use these for supervised learning)
df = df[df["monthly_rent_npr"].notna()].copy()

```

```python
 # anything below this is almost certainly a parsing error
df = df[df["monthly_rent_npr"] >= 3000].copy() 
```

```python
print("CITY:")
print(df["city"].value_counts(dropna=False))

print("\nDISTRICT:")
print(df["district"].value_counts(dropna=False))

print("\nMUNICIPALITY:")
print(df["municipality"].value_counts(dropna=False).head(30))
```

```python
df[df["city"].str.lower().isin(["pokhara", "kaski"])][
    ["listing_id", "location", "municipality", "district", "city"]
]
```

```python
kathmandu_locations = (
    df["city"].str.strip().str.lower().eq("pokhara")
    & df["location"].str.contains(
        "kathmandu|kathmandu metropolis|budhanilkantha|tarkeshwor|nagarjun",
        case=False,
        na=False
    )
)

df.loc[kathmandu_locations, "city"] = "kathmandu"
```

```python
valley_cities = ["kathmandu", "lalitpur", "bhaktapur"]

df_valley = df[
    df["city"]
    .str.strip()
    .str.lower()
    .isin(valley_cities)
].copy()

print("Original rows:", len(df))
print("Kathmandu Valley rows:", len(df_valley))
print("Removed rows:", len(df) - len(df_valley))
```
